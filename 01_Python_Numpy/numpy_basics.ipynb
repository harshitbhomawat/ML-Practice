```python
import numpy as np

#Q1. Create a NumPy array containing integers from 1 to 20.
lst = np.arange(1,21)
print(lst)


#Q2. Create a 3×4 array filled with zeros, and another 4×4 identity matrix.
print(np.full((3,4),0))
print(np.eye(4,4))

#Q3. Given a 3×3 array, print its shape, number of dimensions, total number of elements and data type.
arr = np.full((3,3),1)
print(arr.shape, " ", arr.ndim, " ", arr.size, " ", arr.dtype)

#Q4. Given an array of numbers from 1 to 10, extract all even numbers using Boolean indexing.
lst = np.arange(1,11)
lst = lst[lst%2==0]
print(lst)

#Q5. Given a 3×4 matrix containing numbers from 1 to 12, extract the second row and the third column.
lst = np.reshape(np.arange(1,13), (3,4))
print(lst[1,2])

#Q6. Given an array of marks, calculate its mean, median, standard deviation, maximum and minimum.
lst = np.random.randint(40,100,30)
print(np.mean(lst), " ", np.median(lst), " ", np.std(lst), " ", np.max(lst), " ", np.min(lst))

#Q7. Given a 2D array of shape (3, 4), calculate the sum of every row and every column separately.
arr = np.reshape(np.random.randint(10,100,12),(3,4))
n,m = arr.shape
print("m:",m)

for i in range(n):
    print(np.sum(arr[i]))

for i in range(m):
    print(np.sum(arr[:,i]))

#or

print(np.sum(arr,axis=1)) #row sum
print(np.sum(arr,axis=0)) #column sum


#Q8. Given an array of 12 elements, reshape it into a 3×4 matrix and then transpose it.
arr.T

#Q9. Given an array of prices, apply a 10% discount to every element without using loops.
arr = np.random.randint(400,600,10)
arr = arr*0.9

#Q10. Given two matrices A of shape (2, 3) and B of shape (3, 2), perform matrix multiplication.
a = np.resize(np.arange(10,50,6),(2,3))
b = np.resize(np.arange(20,80,6),(3,2))
a@b

#Q11. Given an array, replace every value below 50 with 0 and keep the remaining values unchanged.
arr = np.random.randint(10,100,20,int)
#approach1:
arr[arr<50] = 0

#approach2:
print(np.where(arr>=50,arr,0))

#Q12. Create a reproducible array of 10 random integers between 1 and 100 using a random seed.
rng = np.random.default_rng(47)
arr1 = rng.integers(1,100,10)
arr2 = rng.integers(1,100,10)
print(arr1)
print(arr2)

```
